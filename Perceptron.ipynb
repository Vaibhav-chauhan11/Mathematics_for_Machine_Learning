```python
import numpy as np
from sklearn.metrics import accuracy_score

T= 1.0
F= 0.0
def get_or_data():
  X =[ [F,F],[F,T],[T,F],[T,T]]
  Y=[[F],[T],[T],[T]]
  return X,Y
X,Y = get_or_data()
# Convert X and Y to numpy arrays
X = np.array(X)
Y = np.array(Y)

class MyPrecepton:
  def __init__(self,learning_rate=0.1,n_iteration=1000):
    self.lr= learning_rate
    self.epochs = n_iteration
    self.weights=None
    self.bias=None

  def fit(self,X,Y):
    self.weights= np.zeros(X.shape[1])
    self.bias=0
    for epoch in range(self.epochs):
      for i in range(X.shape[0]):
        # Corrected typo: Step_activ_func to Step_actuv_function
        Y_pred = self.Step_actuv_function(np.dot(self.weights,X[i])+self.bias)
        mae = Y[i] -Y_pred
        self.weights = self.weights+self.lr *mae*X[i]
        self.bias = self.bias + self.lr*mae

  def Step_actuv_function(self,activation):
    if(activation>=0):
      return 1
    else:
      return 0

  # Added missing predict method
  def predict(self, X):
    predictions = []
    for x in X:
      predictions.append(self.Step_actuv_function(np.dot(self.weights, x) + self.bias))
    return np.array(predictions).reshape(-1, 1)

clf = MyPrecepton()
clf.fit(X,Y)
X_test=[[F,F],[F,T],[T,F],[T,T]]
Y_test =[[F],[T],[T],[T]]
X_test = np.array(X_test)
Y_test = np.array(Y_test)
Y_predicted = clf.predict(X_test)
print(Y_predicted)
# Added missing closing parenthesis
print(accuracy_score(Y_test,Y_predicted))
```

```python
# Multi layer perceptron  , XOR
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential # Corrected import
import tensorflow as tf
import numpy as np # Import numpy
T= 1.0
F= 0.0
def get_XOR_data():
  X =[ [F,F],[F,T],[T,F],[T,T]]
  Y=[[F],[T],[T],[F]]
  return X,Y
X,Y = get_XOR_data()
# Convert X and Y to numpy arrays
X = np.array(X)
Y = np.array(Y)

X_test=[[F,F],[F,T],[T,F],[T,T]]
Y_test =[[T],[F],[T],[F]]
# Convert X_test and Y_test to numpy arrays
X_test = np.array(X_test)
Y_test = np.array(Y_test)

model=Sequential()
model.add(Dense(16,input_shape=(2,),activation='relu'))
model.add(Dense(16,activation='relu')) # Corrected 'dense' to 'Dense'
model.add(Dense(1,activation='sigmoid')) # Corrected 'dense' to 'Dense'
model.compile(loss='binary_crossentropy',optimizer='SGD',metrics=['accuracy'])
model.fit(X,Y,epochs=100) # Changed epoch to epochs
loss,accuracy=model.evaluate(X_test,Y_test,verbose=0) # Corrected 'evalute' to 'evaluate'
print('Accuracy: % 2f' % (accuracy*100))
print(loss)
```
