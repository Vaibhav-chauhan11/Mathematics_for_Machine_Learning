```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# ------------------------------------------
# 1. Create 20 Sample Points (2D)
# ------------------------------------------
data = np.array([
    [1, 2], [2, 1], [1.5, 1.8], [3, 3], [2.5, 2.8],
    [8, 8], [9, 9], [8.5, 8.2], [9.1, 8.8], [8.2, 9.3],
    [4, 10], [5, 11], [4.5, 10.5], [5.2, 10.8], [4.8, 11.3],
    [0, -1], [1, -2], [0.5, -1.5], [2, -3], [1.8, -2.5]
])

# ------------------------------------------
# 2. Apply K-Means
# ------------------------------------------
k = 3  # choose 3 clusters
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(data)
print(kmeans.labels_)

labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# ------------------------------------------
# 3. Plot Results
# ------------------------------------------
plt.scatter(data[:, 0], data[:, 1], c=labels, s=50)
plt.scatter(centroids[:, 0], centroids[:, 1], c='black', s=150, marker='X')

plt.title("K-Means Clustering on 20 Sample Points")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```
